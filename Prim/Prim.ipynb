## Prim's Algorithm for calculation of a minimal spanning tree

```python
import math
```

## The efficient implementation is based on a min priority queue, implemented as min heap

```python
# note: this is not a complete implementation
# - create heap not required, it is here a simple array creation
# contains nodes to be added to minimal spanning tree
# the key for the priority is given in an array, indexed by the priority queue values
```

```python
# internal functions for navigating up and down in the heap
def left_child(i):
   return 2*i + 1
def right_child(i):
   return 2*i + 2
def parent(i):
   return (i-1) // 2
```

```python
# internal function for reestablishing the min heap
def min_heapify(i, value, pos):
   left = left_child(i)
   right = right_child(i)
   if left < heapsize and value[a[left]] < value[a[i]]:
      minimum = left
   else:
      minimum = i
   if right < heapsize and value[a[right]] < value[a[minimum]]:
      minimum = right
   if minimum != i:
      pos[a[i]], pos[a[minimum]] = pos[a[minimum]], pos[a[i]]
      a[i], a[minimum] = a[minimum], a[i]
      min_heapify(minimum, value, pos)
```

```python
# external function for extracting the minimum - used by prim()
def extract_min(value, pos):
   global heapsize
   if heapsize > 0:
      min_elem = a[0]
      heapsize -= 1
      pos[a[heapsize]] = pos[a[0]]
      pos[a[0]] = False
      a[0] = a[heapsize]
      min_heapify(0, value, pos)
      return min_elem
   else:
      return None
```

```python
# external function for decreasing the cost - used by prim()
def decrease_value(elem, new_value, value, pos):
   i = pos[elem]
   if i >= 0 and i < heapsize and new_value < value[a[i]]:
      value[a[i]] = new_value
      while i > 0 and value[a[i]] < value[a[parent(i)]]:
         pos[a[i]], pos[a[parent(i)]] = pos[a[parent(i)]], pos[a[i]]
         a[i], a[parent(i)] = a[parent(i)], a[i]
         i = parent(i)
      return True
   else:
      return False
```

## Prim's algorithm

### Internal data for Prim's algorithm

```python
# number of nodes in graph g
n = -1

# heapsize for the min priority queue
heapsize = 0
# array with elements for the min priority queue
a = []
# position / index of element in the min priority queue
pos = []

# predecessor of discovered node
pred = []
# set of unvisited nodes
unvisited = []
# costs to predecessor node, used as values for the min priority queue
cost = []
```

### Initialization of data, to be called before every call of Prim's algorithm

```python
def initialization():
   global heapsize, a, pos, pred, unvisited, cost, n
   n = len(g)
   heapsize = n
   a = [i for i in range(heapsize)]
   pos = [i for i in range(heapsize)] # initial state identical to a
   # predecessor of node and shortest cost from a visited node at current point in time
   pred = [-1]*n
   unvisited = [True for _ in range(n)]
   cost = [math.inf]*n
   # starting node 0
   cost[0] = 0
```

### Prim's algorithm

```python
# with parameter True the algorithms outputs data in every iteration
def prim(outputInternalStates = False):
   initialization()
   # output of internal states
   if outputInternalStates:
      print(cost, pred, unvisited)
   # while there are unvisited reachable nodes
   # note: infinite loop if graph is not connected, i. e. there are unreachable nodes from the starting node
   while heapsize > 0:
      # current edge with minimal cost from a visited node to an unvisted node
      node_min_cost = extract_min(cost, pos)
      unvisited[node_min_cost] = False
      # update cost to all adjacent / reachable nodes
      for edge in g[node_min_cost]:
         # new minimal cost (edge[1]) to node (edge[0]) found?
         if unvisited[edge[0]] and edge[1] < cost[edge[0]]:
            pred[edge[0]] = node_min_cost
            decrease_value(edge[0], edge[1], cost, pos)
      # output of internal states
      if outputInternalStates:
         print(cost, pred, unvisited)
```

### Output of Prim's algorithm

```python
# outputs the minimal spanning tree with node numbers
def print_mst():
   print("Overall cost =", sum(cost))
   print("Edges:")
   for i in range(1,n):
      print(i, "<->", pred[i], "cost =", cost[i])
```

### Output of Prim's algorithm

```python
# outputs the minimal spanning tree with node letter, starting with parameter first_node
def print_mst_node_name(first_node):
   print("Overall cost =", sum(cost))
   print("Edges:")
   for i in range(1,n):
      print(min(chr(i+ord(first_node)),chr(pred[i]+ord(first_node))), "<->", \
            max(chr(i+ord(first_node)),chr(pred[i]+ord(first_node))), "cost =", cost[i])
```

## Graph needs to be symmetric given as adjacency list

### This function completes the adjacency list to a symmetric graph

```python
def complete(g):
   for i in range(len(g)):
      for j in range(len(g[i])):
         if (i,g[i][j][1]) not in g[g[i][j][0]]:
            g[g[i][j][0]].append((i,g[i][j][1]))
```
