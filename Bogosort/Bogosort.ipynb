## Sorting an array using bogosort

```python
import random
```

```python
# checks whether array is sorted correctly, this also works with identical keys in the array
def sorted(a):
    return all(a[i] <= a[i+1] for i in range(len(a)-1))
```

```python
def bogosort(a):
    while not sorted(a):
        random.shuffle(a)
    return a
```

```python
# this variant returns also the number of shuffle operations
def bogosort_num_shuffle(a):
    num_shuffle = 0
    while not sorted(a):
        random.shuffle(a)
        num_shuffle += 1
    return a, num_shuffle
```

```python
# checks whether array is sorted correctly, this also works with identical keys in the array
def check_sorting(a):
    return sorted(a)
```

## Constants

```python
# The size of array in tests can be changed here - better not too large ...
ARRAY_SIZE = 10
# Some output text
TEXT_BEFORE_SORTING = "Before sorting:"
TEXT_AFTER_SORTING = "After  sorting:"
TEXT_CHECK_SORTING = "Sorting correct? -"
TEXT_SHUFFLE_NUM = "Number of shuffle operations:"
```

## Generate test data and test algorithm

### Sort array with bogosort

```python
# test sorting sorted array
sorted_array = list(range(ARRAY_SIZE))
print(TEXT_BEFORE_SORTING, sorted_array)
a = bogosort(sorted_array)
print(TEXT_AFTER_SORTING, a)
print(TEXT_CHECK_SORTING, check_sorting(a))
```

```python
# test sorting reversly sorted array
reverse_sorted_array = list(range(ARRAY_SIZE-1, -1, -1))
print(TEXT_BEFORE_SORTING, reverse_sorted_array)
a = bogosort(reverse_sorted_array)
print(TEXT_AFTER_SORTING, a)
print(TEXT_CHECK_SORTING, check_sorting(a))
```

```python
# test sorting random array
# use numpy for generating a permutation
import numpy
# initialialize random number generator to obtain reproducable results
# other values might be used or even no initialization done here
numpy.random.seed(0)
# now generate random data and test sorting
array = [int(x) for x in numpy.random.permutation(ARRAY_SIZE)]
print(TEXT_BEFORE_SORTING, array)
a = bogosort(array)
print(TEXT_AFTER_SORTING, a)
print(TEXT_CHECK_SORTING, check_sorting(a))
```

### Sort array with bogosort and output the number of shuffle operations

```python
# test sorting sorted array
sorted_array = list(range(ARRAY_SIZE))
print(TEXT_BEFORE_SORTING, sorted_array)
a = bogosort_num_shuffle(sorted_array)
print(TEXT_AFTER_SORTING, a[0])
print(TEXT_SHUFFLE_NUM, a[1])
print(TEXT_CHECK_SORTING, check_sorting(a[0]))
```

```python
# test sorting reversly sorted array
reverse_sorted_array = list(range(ARRAY_SIZE-1, -1, -1))
print(TEXT_BEFORE_SORTING, reverse_sorted_array)
a = bogosort_num_shuffle(reverse_sorted_array)
print(TEXT_AFTER_SORTING, a[0])
print(TEXT_SHUFFLE_NUM, a[1])
print(TEXT_CHECK_SORTING, check_sorting(a[0]))
```

```python
# test sorting random array
# use numpy for generating a permutation
import numpy
# initialialize random number generator to obtain reproducable results
# other values might be used or even no initialization done here
numpy.random.seed(0)
# now generate random data and test sorting
array = [int(x) for x in numpy.random.permutation(ARRAY_SIZE)]
print(TEXT_BEFORE_SORTING, array)
a = bogosort_num_shuffle(array)
print(TEXT_AFTER_SORTING, a[0])
print(TEXT_SHUFFLE_NUM, a[1])
print(TEXT_CHECK_SORTING, check_sorting(a[0]))
```
