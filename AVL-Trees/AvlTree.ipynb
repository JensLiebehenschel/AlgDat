## AVL Trees with textual visualization

## The implementation is inspired by M. Nebel: Entwurf und Analyse von Algorithmen, 2012, p. 130-140

## Data structure for storing nodes of an AVL tree

```python
class Node:
   def __init__(self, data):
      self.data = data
      self.left = None
      self.right = None
      self.balance = 0
```

## Insert including balancing function and search function

```python
def insert(node, data):
   curr_height_increased = False
   if not node:
      node = Node(data)
      curr_height_increased = True
   elif data < node.data:
      (node.left, child_height_increased) = insert(node.left, data)
      if child_height_increased:
         if node.balance == 1:
            if node.left.balance == 1: # right rotation
               root = node.left
               node.left, root.right, node = root.right, node, root
               node.balance, node.right.balance = 0, 0
            else: # left-right rotation
               root = node.left.right
               node.left.right, root.left = root.left, node.left
               node.left, root.right, node = root.right, node, root
               node.left.balance = 1 if node.balance == -1 else 0
               node.right.balance = -1 if node.balance == 1 else 0
               node.balance = 0
         else: # node.balance in [-1,0]:
            node.balance += 1
            if node.balance == 1:
               curr_height_increased = True
   else: # data >= node.data
      (node.right, child_height_increased) = insert(node.right, data)
      if child_height_increased:
         if node.balance == -1:
            if node.right.balance == -1: # left rotation
               root = node.right
               node.right, root.left, node = root.left, node, root
               node.balance, node.left.balance = 0, 0
            else: # right-left rotation
               root = node.right.left
               node.right.left, root.right = root.right, node.right
               node.right, root.left, node = root.left, node, root
               node.left.balance = 1 if node.balance == -1 else 0
               node.right.balance = -1 if node.balance == 1 else 0
               node.balance = 0
         else: # node.balance in [0,1]:
            node.balance -= 1
            if node.balance == -1:
               curr_height_increased = True
   return (node, curr_height_increased)
```

```python
def search(node, data):
   if not node or data == node.data:
      return node # not found or found
   elif data < node.data:
      return search(node.left, data)
   else: # data >= node.data
      return search(node.right, data)
```

## Supporting output function for testing

```python
def inorder(node, level=0):
    if node:
        inorder(node.left, level+1)
        print(level, "  "*level, "%s%s"%(node.data, "[" + str(node.balance) + "]"))
        inorder(node.right, level+1)
```

## First test

```python
# keys to be inserted
keys = [7,3,9,1,5]
# this is the resulting tree
#    7
#  3   9
# 1 5
```

```python
# create empty AVL tree
root = None

# insert keys
for k in keys:
    root, _ = insert(root, k)
keys.sort()
print("Inserted keys:", keys)
```

```python
# output the inorder of the tree
print("Inorder of tree:")
inorder(root)
```

## Test with a larger tree - no rotation required

```python
# keys to be inserted
keys = [3,6,2,4,8,1,5,7,9]
# this is the resulting tree
#    3
#  2     6
# 1    4   8
#       5 7 9
# additional keys not in the tree
keys_ext = keys + [0,10]
```

```python
# create empty AVL tree
root = None

# insert keys
for k in keys:
    root, _ = insert(root, k)
keys.sort()
print("Inserted keys:", keys)
```

```python
# output the inorder of the tree
print("Inorder of tree:")
inorder(root)
```

```python
# search values
keys_ext.sort()
for k in keys_ext:
    res = search(root, k)
    if res:
        print(k, "found:", res.data)
    else:
        print(k, "not found")
```
