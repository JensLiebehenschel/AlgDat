## Traveling Salesperson Algorithm

```python
import itertools
import math
```

```python
# x is infinity for easier handling of adjacency matrix of graph
inf = math.inf
x = -inf # no path from ... to ... in adjacency matrix
```

## Output of the graph's adjacency matrix

```python
def output(g):
   for i in range(len(g)):
      for j in range(len(g)):
         print(str(g[i][j])+" " if g[i][j] >= 0 else "x ", end="")
      print()
```

## Main algorithm for calculation of the shortest tour

```python
# we assume the starting (and ending) node is at index 0 of g
def traveling_salesperson(g):
   min_cost = inf
   min_cycle = None
   for p in itertools.permutations(range(1, len(g))):
      current_cost = 0
      # check if the cycle described by p is closed
      cycle = (0,) + p + (0,)
      valid_cycle = True
      i = 0
      while valid_cycle and i < len(cycle)-1:
         if g[cycle[i]][cycle[i+1]] >= 0:
             current_cost += g[cycle[i]][cycle[i+1]]
         else:
            valid_cycle = False
         i += 1
      # check whether current cycle is shorter
      if (valid_cycle and current_cost < min_cost):
         min_cost = current_cost
         min_cycle = cycle
         # output new shortest tour 
         print("New shortest tour:", min_cycle, ", length:", min_cost)
   return (min_cycle, min_cost)
```

## Graph given as adjacency matrix
It might be directed or undirected (symmetric in the latter case)

### First test

```python
g = [[0, 5, 7, x, 3, 6, x, x],
     [5, 0, x, 4, x, 3, 2, x],
     [7, x, 0, x, 1, x, x, 2],
     [x, 4, x, 0, 2, x, 1, x],
     [3, x, 1, 2, 0, x, 3, 4],
     [6, 3, x, x, x, 0, 1, 1],
     [x, 2, x, 1, 3, 1, 0, 2],
     [x, x, 2, x, 4, 1, 2, 0]
    ]
```

```python
output(g)
```

```python
print("Result of algorithm:", traveling_salesperson(g))
```

### Another test with a graph that leads to several new shortest tours if the permutations are generated lexicographically
